# Natural frequencies of circular cylindrical shells: Leissa (1973)

Circular cylindrical shells and open panels supported by shear diaphragms (SD) on all edges, with the CLPT models
`'cylshell_clpt_donnell'` and `'cylshell_clpt_sanders'` and the FSDT and TSDT models with Sanders' kinematics,
`'cylshell_fsdt_sanders'` and `'cylshell_tsdt_sanders'`, against:

- the frequencies of 3D elasticity of Leissa [1], Table 2.8, and the percent differences of the shell theories of
  Tables 2.6 and 2.7;
- the deep open shell of Table 2.46 of [1], $\theta_0 = 270^\circ$;
- Love's theory of Loy, Lam and Shu [2], as reproduced in Table 1 of Li, Fu and Batra [3];
- the exact Navier solutions of the same kinematics, which Leissa's tables confirm to the printed digits.

The frequency parameter is $\Omega = \omega R \sqrt{\rho (1 - \nu^2)/E}$, with $E = 70$ GPa, $\nu = 0.3$ and
$\rho = 2700$ kg/m$^3$; $R/h$, $L/R$ and $n$ define each case.

According to Leissa [1] (p. 158), an open panel of angle $\theta_0$ supported on all four edges by shear diaphragms
has the same frequencies as the closed shell with $n = \pi/\theta_0$ circumferential waves: each closed shell is
modelled here by the panel of angle $\pi/n$. Leissa's values of the thin-shell theories neglect the rotary inertia,
which the mass matrices of panels include: for the CLPT models it is removed exactly from two mass matrices, since
the rotary inertia scales with $h^3$ and the translational inertia with $h$: `calc_kM()` with $(h, \rho)$ and
$(2h, \rho/2)$ gives $M_1 = M_t + M_r$ and $M_2 = M_t + 4 M_r$, so that $M_t = (4 M_1 - M_2)/3$. The FSDT and TSDT
models keep the rotary inertia and the transverse shear deformation, as 3D elasticity does.

All strains of Sanders' theory vanish for rigid-body motions [4], which is not the case for Love's or Donnell's
theories.

The models, the SD boundary conditions, the mode identification (projection onto
$\sin(\pi x/L)\sin(\pi y/b)$) and the Navier solutions are those of the test modules
`tests/tests_shell/test_sanders_kinematics.py` and `tests/tests_shell/test_fsdt_tsdt_cylinders.py`, imported here.

```python
import os
import sys
import time

import numpy as np
from IPython.display import Markdown, display

sys.path.insert(0, os.path.abspath(os.path.join('..', 'tests', 'tests_shell')))
import test_sanders_kinematics as sk
import test_fsdt_tsdt_cylinders as ft


def pct(a, b):
    return 100*(a/b - 1)


def markdown_table(header, rows):
    lines = ['| ' + ' | '.join(header) + ' |', '|' + '---|'*len(header)]
    lines += ['| ' + ' | '.join(str(c) for c in row) + ' |' for row in rows]
    return '\n'.join(lines)
```

## 1. Tables 2.6 and 2.8: Donnell and Sanders kinematics, CLPT

Panels of angle $\pi/n$ with $12 \times 12$ terms (`sk.make_panel`), the frequency parameter of the mode with one
half-wave in each direction (`sk.frequency_parameter`), against the Navier solution of the same kinematics
(`sk.navier_Omega`) and the reference value: 3D elasticity of Table 2.8 of [1] for $R/h = 20$, and Love's theory
of [2] for $R/h = 100$, $L/R = 20$, $n = 2$. Differences in percent.

```python
# (R/h, L/R, n, reference, source); the reference values of Leissa (1973), Table 2.8, and Loy et al. (1997)
CASES = [(20, 4, 2, 0.121249, 'Leissa (1973), T2.8, 3D elasticity'),
         (20, 20, 1, 0.0161063, 'Leissa (1973), T2.8, 3D elasticity'),
         (20, 4, 3, 0.129881, 'Leissa (1973), T2.8, 3D elasticity'),
         (100, 20, 2, 0.009382, 'Loy et al. (1997), Love')]
R = 1.
rows = []
t0 = time.time()
for Rh, LR, n, ref, source in CASES:
    h, L = R/Rh, LR*R
    row = [Rh, LR, n, '%.6g' % ref, source]
    for kin in ('donnell', 'sanders'):
        s = sk.make_panel('cylshell_clpt_' + kin, R, h, L, np.pi/n)
        Om = sk.frequency_parameter(s)
        On = sk.navier_Omega(kin, R, h, L, n)
        row += ['%.6g (%+.3f%%)' % (On, pct(On, ref)), '%.6g (%+.4f%%)' % (Om, pct(Om, On))]
    rows.append(row)
print('%.0f s' % (time.time() - t0))
display(Markdown(markdown_table(['$R/h$', '$L/R$', '$n$', 'reference', 'source',
                                 'Navier Donnell (vs ref.)', 'panels Donnell (vs Navier)',
                                 'Navier Sanders (vs ref.)', 'panels Sanders (vs Navier)'], rows)))
```

## 2. Table 2.46: deep open shell, $\theta_0 = 270^\circ$

$R/h = 20$, $L/R = 100$, an open shell of $\theta_0 = 3\pi/2$ with SD on all edges. Donnell's value of Table 2.46
of [1] is 0.00374949. Sanders' theory is compared with Flügge's value of the same table, 0.00459872, which is 0.1 %
below the exact Navier solution of Sanders' kinematics, 0.00460365 (computed with the Navier solution of
`test_sanders_kinematics.py`, not printed in [1]).

```python
h, L = R/20, 100*R
rows = []
for model, ref, label in (('cylshell_clpt_donnell', 0.00374949, 'Donnell, Table 2.46'),
                          ('cylshell_clpt_sanders', 0.00459872, 'Flügge, Table 2.46')):
    s = sk.make_panel(model, R, h, L, 3*np.pi/2)
    Om = sk.frequency_parameter(s)
    rows.append([model, '%.6g' % ref, label, '%.6g (%+.2f%%)' % (Om, pct(Om, ref))])
display(Markdown(markdown_table(['model', 'reference', 'source', 'panels'], rows)))
```

## 3. Table 2.8: short waves, $R/h = 500$

$R/h = 500$, $n = 4$, $L/R = 1$: 3D elasticity gives $\Omega = 0.354118$, Donnell's theory 0.354145 and Sanders'
0.35412 [1]: for short wavelengths Donnell's kinematics are as accurate as Sanders'.

```python
h, L, n = R/500, R, 4
rows = []
for model, ref in (('cylshell_clpt_donnell', 0.354145), ('cylshell_clpt_sanders', 0.35412)):
    s = sk.make_panel(model, R, h, L, np.pi/n)
    Om = sk.frequency_parameter(s)
    rows.append([model, '%.6g' % ref, '%.6g (%+.4f%%)' % (Om, pct(Om, ref)), '%+.4f%%' % pct(Om, 0.354118)])
display(Markdown(markdown_table(['model', 'Leissa (1973), T2.8, same theory', 'panels', 'panels vs 3D elasticity'],
                                rows)))
```

## 4. Table 2.8 with the FSDT and the TSDT

The cases of $R/h = 20$ of Section 1 with `'cylshell_fsdt_sanders'` and `'cylshell_tsdt_sanders'` ($14 \times 14$
terms, `ft.ritz_frequency`), with the rotary inertia and the transverse shear deformation, against 3D elasticity
and against the Navier solution of the same model (`ft.navier`), and the CLPT with Sanders' kinematics of
Section 1.

```python
factor = R*np.sqrt(sk.RHO*(1 - sk.NU**2)/sk.E)
rows = []
t0 = time.time()
for Rh, LR, n, ref, source in CASES[:3]:
    h, L = R/Rh, LR*R
    row = [Rh, LR, n, '%.6g' % ref]
    for model in ('cylshell_fsdt_sanders', 'cylshell_tsdt_sanders'):
        s = ft.make_panel(model, R, h, L, np.pi/n, m=14, n=14)
        Om = ft.ritz_frequency(s)*factor
        On = ft.navier(model, R, h, L, n, s.ABD, sk.RHO)*factor
        row += ['%.6g (%+.3f%%)' % (Om, pct(Om, ref)), '%+.1e' % (Om/On - 1)]
    clpt = sk.make_panel('cylshell_clpt_sanders', R, h, L, np.pi/n, m=14, n=14)
    Oc = ft.ritz_frequency(clpt)*factor
    row.append('%.6g (%+.3f%%)' % (Oc, pct(Oc, ref)))
    rows.append(row)
print('%.0f s' % (time.time() - t0))
display(Markdown(markdown_table(['$R/h$', '$L/R$', '$n$', '3D elasticity, T2.8', 'panels FSDT', 'FSDT vs Navier',
                                 'panels TSDT', 'TSDT vs Navier', 'panels CLPT Sanders (with rotary inertia)'],
                                rows)))
```

## Discussion

- **Ritz against Navier.** The panels of angle $\pi/n$ reproduce the Navier solutions of the same kinematics to
  better than $10^{-6}$, the closed shell with $n$ waves of [1] (p. 158).
- **Donnell against Sanders.** For long shells with few circumferential waves Donnell's kinematics overestimate the
  frequency, by 18.9 % for $R/h = 20$, $L/R = 20$, $n = 1$, whereas Sanders' kinematics are within 0.021 % of 3D
  elasticity there; for the deep open shell of Table 2.46 Donnell's theory is 18 % below Flügge's value. For short
  waves, $R/h = 500$, $n = 4$, $L/R = 1$, both are within 0.01 % of 3D elasticity.
- **Transverse shear and rotary inertia.** With them, the FSDT and TSDT with Sanders' kinematics are within 0.07 %
  of the 3D elasticity solution of Table 2.8 for $R/h = 20$; for $L/R = 4$ and $n = 3$, where the CLPT with Sanders'
  kinematics is 0.21 % above it, they are three times closer.
